# **<span style="color:yellow">Part B<span>** - Working with the Iris Dataset

### **1.** Load the Iris dataset

````python
# First import the dataset then load it

from sklearn.datasets import load_iris

iris = load_iris()
````

`load_iris()` is a built-in function in Scikit-learn.

The Iris dataset contains 150 flower samples from 3 different species.

### **2.** Print the number of rows and columns

````python
print("Number of rows: ", iris.data.shape[0])

print("Number of columns: ", iris.data.shape[1])
````

* **150** = number of rows / samples / instances

* **4** = number of features / attributes / columns

### **3.** Print feature names

````python
print("Feature Names: ")
print(iris.feature_names)
````

### **4.** Print target names

````python
print("Target Names: ")
print(iris.target_names)
````

### **5.** Convert the dataset into a Pandas DataFrame

````python
df = pd.DataFrame(iris.data, columns=iris.feature_names)

# Displaying it
df
````

>At this point, the DataFrame contains only the features.

Now let's add the target column.

````python
df["target"] = iris.target

# Displaying it
df
````

### **6.** Add the actual flower species names

The target values are numbers:

<center>

0

1

2

</center>

We can convert them into species names:

````python
df["species"] = iris.target_names[iris.target]

# To display 1st 5 rows of the dataset
df.head()
````

> **i.** Structured data
>

### Identify each example

| Example                                | Data Type       | Why?                                                           |
| -------------------------------------- | --------------- | -------------------------------------------------------------- |
| Customer database in MySQL             | Structured      | Data is organized into rows and columns                        |

### Which type is easiest for traditional ML?

> Structured data

### Why?

> Structured data is already organized into rows and columns, making it easier to use as input for traditional machine learning algorithms.

### Why is unstructured data more challenging?

> Unstructured data does not follow a predefined format. Therefore, additional processing and feature extraction are usually required before it can be used effectively by machine learning algorithms.
>
> **Examples:**
> Images, Videos, Audio, Free text

### **3.** Underfitting, Good Fit and Overfitting

You are given:

| Model | Training Accuracy | Testing Accuracy |
| ----- | ----------------: | ---------------: |
| A     |               99% |              61% |
| B     |               90% |              88% |
| C     |               58% |              55% |

##### **Model A** - Overfitting

**Why?** The training accuracy is very high, but the testing accuracy is much lower.

<div align=center>

```python
Training: 99%
Testing:  61%
```

</div>

* The model learned the training data too specifically and cannot generalize well to unseen data.

##### **Model B** - Good Fit

<div align=center>

```python
Training: 90%
Testing:  88%
```

</div>

* Both accuracies are high and close to each other.

* This means the model generalizes reasonably well.

##### **Model C** - Underfitting

<div align=center>

```python
Training: 58%
Testing:  55%
```

</div>

* Both accuracies are low.

* This means the model has not learned the patterns in the data sufficiently.

- - -

<div align=center style="color:lightblue">

#### <u>**Main disadvantage of overfitting**</u>
</div>

An overfitted model performs very well on the training data but poorly on new or unseen data.

<div align=center style="color:lightpink">

#### <u>**Techniques to reduce overfitting**</u>
</div>

* Cross-validation
* Regularization
* Reducing model complexity
* Feature selection
* Early stopping
* Using more training data

<U> **In Simple:** </u> Regularization can help reduce overfitting by preventing the model from becoming too complex.

### **4.** Training, Validation and Testing Sets

<div align="left">

#### <span style="color:lightgreen">**Training Set**</span>

***Used to:*** Train the machine learning model and learn patterns from the data.

</div>

<div align="center">

#### <span style="color:lightgreen">**Validation Set**</span>

***Used to:*** Tune the model and select the best hyperparameters during development.

</div>

<div align="right">

#### <span style="color:lightgreen">**Testing Set**</span>

***Used to:*** Evaluate the final performance of the model using unseen data.

</div>

#### **●** Why use a separate test dataset?

> A separate test dataset provides an unbiased evaluation of how well the model can generalize to unseen data.

#### **●** A student trains and evaluates a model using the same dataset and gets 99% accuracy. Does this mean the model performs well?

>No.
>
>The model may have memorized or overfitted the training data. High accuracy on the same dataset does not guarantee good performance on unseen data. A separate test dataset is required to evaluate the model's ability to generalize.
