## **Part D – Understanding the concepts**

### 1. Why is data understanding essential before training a machine learning model? 

> Data understanding is essential because it helps identify data quality issues (like missing values, outliers, and duplicates) and reveals relationships between variables. This knowledge is crucial for selecting appropriate preprocessing steps and algorithms, which ultimately leads to better model performance.


---

### 2. What is the difference between univariate and bivariate analysis?

> - **Univariate analysis** involves examining a single variable at a time to understand its distribution (e.g., using histograms or boxplots).
> - **Bivariate analysis** involves examining two variables simultaneously to understand the relationship or correlation between them (e.g., using scatter plots or correlation matrices).


---

### 3. Why should duplicate records be removed?

> Duplicate records can bias the model by giving undue importance to certain instances, leading to overfitting. They can also cause data leakage if the same sample appears in both the training and testing sets.


---

### 4. Why can missing values reduce model performance? 

> Missing values can reduce model performance because they introduce bias and reduce the representativeness of the dataset. Moreover, most machine learning algorithms cannot handle missing data directly and will throw errors during model training.


---

### 5. What is data leakage? Give one example. 

> **Data leakage** occurs when information from outside the training dataset (such as the test set) is inadvertently used to create the model, leading to artificially high performance during testing that will not generalize.
>
> **Example:** Scaling or imputing the entire dataset (training + testing) before splitting it.


---

### 6. Why must train-test splitting be performed before preprocessing?

> Train-test splitting must be performed before preprocessing to prevent **data leakage**. If preprocessing is done on the entire dataset before splitting, the parameters will be influenced by the test set, making the test evaluation invalid.
